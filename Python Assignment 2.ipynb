```python
def calculate_avg(*marks):
    if not marks:
        return "Error: No Marks Provided"
    for ex in marks:
        if not isinstance(ex,(int,float)) or isinstance(ex, bool):
            return "Error: Marks Must be Numaric"
        if ex<0 or ex>100:
            return "Error: Marks must be between 0 and 100"
        return sum(marks)/len(marks)
        
print(calculate_avg(55,45,58,97.0, 87,67))
print(calculate_avg())
print(calculate_avg(45,76,True))
```

```python
def CalaculateBill(price, discount=0):
    discount_amount=price*discount/100
    bill=price-discount_amount
    
    print("Original Price: ",price)
    print("Discount : ",discount, "%")
    print("Discount Amount: ",discount_amount)
    print("Final Bill Amount: ",bill)
    print(" ===================== ")
    
CalaculateBill(20000)# Cash 0% discount
CalaculateBill(10000,10)# UPI 10% discount
CalaculateBill(30000,20)# Card 20% discount
```

```python
#Text to Image >>>> (Image, ImageDraw, ImageFont)
from PIL import Image, ImageDraw, ImageFont

txt=input("Enter Your Text: ")

#Create Image
image=Image.new("RGB",(800,300), "white")

#create drawing Object
draw=ImageDraw.Draw(image)

#Load Font
font= ImageFont. load_default() # fontsize, fontStyle,

#Set left and Right Margins
left_margin=50
right_margin=50


#Calculate Width
available_width=800-left_margin-right_margin

#Break text into multiline
words=text.split()
wrapped_lines=[]
current_line=""

for word in words:
    test_line=current_line + " "+word if current_line else word
    text_width= draw.textbbox((0,0), test_line, font=font) [2]
    
    if text_width <= available_width:
        current_line=test_line
    else:
        if current_line:
            wrapped_lines. append(current_line)
        current_line=word
        
if current_line:
    wrapped_lines.append(current_line)
    
#Draw text Line by Line
y=50

for line in wrapped_lines:
    draw. text((left_margin,y), line, fill="red", font=font)
    y+=25
image.save("test_sample_24.png")
print("Image Generated Successfully!")
```

```python
# args
def add(*args):  # tuple
    total = 0
    for ex in args:
        total = total + ex
    return total

print(add(10, 20))
print(add(20, 30, 40))
print(add(20, 30, 40, 50))
print(add(20, 30, 40, 50, 60))
```

```python
# args
def add(*args): # tuple
    total=0 # 30
    for ex in args:
        total=total+ex # 150
    print("Sum: ", total)
    
add(10,20) # args=(10,20)
add(40,50,60) # args=(40,50,60)
add(20,30,40,50) # args=(20,30,40,50)
add(20,30,40,50,60) # args=(20,30,40,50,60)
```

```python
#Excercise 4:
def studentResult(name, *marks, ** details):
    print("Student Name: ", name)
    print("Subject Marks: ",marks)
    
    if len(marks)>0:#false
        total=sum(marks)
        avg=total/len(marks)
        print("Total Marks: ",total)
        print("Avg Marks: ",avg)
    print(" ========= Additional Information ======= ")

    for key, value in details.items():
        print(key, ":", value)
studentResult("Amit",50,80,49,66,97, roll=101, Sem="1ST")
```
