```python
# Student records
students = [
    {"name": "Aman", "marks": [80, 75, 90]},
    {"name": "Riya", "marks": [35, 40, 30]},
    {"name": "Rahul", "marks": [90, 85, 95]},
    {"name": "Neha", "marks": [60, 70, 65]}
         
]
 
def generate_result(student_record):
    for student in student_record:
        name=student['name'] #Aman
        marks=student['marks'] #[80, 75, 90]

        
        avg=sum(marks) /len(marks)#
        if avg>=40:
            status="Pass"
    else:
        status="Fail"
    yield {
        "name":name,
        "average":avg,
        "status": status
    }
    
for result in generate_result(students):
    print(result['name' ])
    print(result['average'])
    print(result['status'])
    print("-"*25)
```

```python
class Employee:
    def addEmpDetails(self,E_ID, E_Name, E_Salary,E_Company):
        self.id=E_ID #(E_ID is a local variables)
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company
        #self.company is an instance variable
 
    def showEmpDetails(self):
        print("Employee ID: ",self.id)
        print("Employee Name: ",self.name)
        print("Employee Salary: ", self.salary)
        print("Employee Company: ",self.company)
        
e1=Employee()
e1.addEmpDetails(111,"Raj",12000.00,"HCL")
e1.showEmpDetails()
print("="*25)

e2=Employee()
e2.addEmpDetails(222,"Ansh Chaudhary",22000.00,"HCL")
e2.showEmpDetails()
print("="*25)
```
